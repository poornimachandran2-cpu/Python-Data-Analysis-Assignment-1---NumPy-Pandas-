## ***Numpy Array Operations:***

### 1. Create a 1D NumPy Array:

NumPy (Numerical Python) is a powerful library for
numerical computations.

```python
import numpy as np
```

```python

# Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values:
# [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
temperatures_w1 =np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 temperature: ")
print(temperatures_w1)
```

*  np.array : creates NumPy array




```python
# 2.Inspection and Properties:
# Inspect the shape, data type, and the number of elements in the array. Inspect the shape of the temperatures_w1 array.
temperatures_w1
print(type(temperatures_w1))
print(" Shape: ", temperatures_w1.shape)
print(" No of elements: ",temperatures_w1.size)
print(" Data type: ",temperatures_w1.dtype)
```

* (.shape) → tuple of dimensions
* (.size) → total number of elements
* (.dtype) → data type of elements

```python
# 3. Array Operations:
# Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32
fahrenheit = (temperatures_w1 * 9/5) + 32
print(fahrenheit)
```

•Allows operations between arrays of different shapes.

```python
# Find the maximum, minimum, and mean temperatures for the week.
print("Maximum Temperature: ",np.max(fahrenheit))
print("Minimum Temperature: ",np.min(fahrenheit))
print("Mean Temperature: ",np.mean(fahrenheit))
```

```python
# 4. Array Slicing and Indexing:
# Extract the temperatures for the first three days of the week.
print(temperatures_w1[:3])

# Extract the temperatures for the weekend (last two days).
print(temperatures_w1[-2:])

# Extract the temperatures for the middle three days of the week.
print(temperatures_w1[2:5])
```

*  Starts from index 0,Stops before index 3.
*  Extracts last 2 days.
*  Extracts Middle 3 days.

```python
# 5. Create a 2D Array:
# Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
# Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
# Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

2D array shows the 2 dimensional array object with 2 rows and 7 columns.

```python
# 6. Inspect and Slice the 2D Array:
# Inspect and print the shape, data type, and total number of elements in the temperatures array.
print(type(temperatures))
print("Shape: ",temperatures.shape)
print("No of elements: ",temperatures.size)
print("Data type: ",temperatures.dtype)
```

* (.shape) → tuple of dimensions
* (.size) → total number of elements
* (.dtype) → data type of elements

```python
# Extract the temperatures for each week and the weekends (last two days) for both weeks.

print(temperatures[:, -2:])
```

* : → Select all rows.
* -2: → Select the last two columns.
* Returns the temperatures for the last two days (weekend) of both weeks.

## ***Pandas Series:***

```python
import pandas as pd
```

A one-dimensional labeled array capable of
holding any data type (int, float, string, Python objects).

```python
# 1. Creating Pandas Series:
# Create a Pandas Series named marks with the following data:
# Marks: 95, 92, 89, 85, 80
# Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
# 2. Indexing and Slicing:
#  Use the integer index position to access the mark of the 1st rank student.
print(marks["Rank1"])
```

access the mark of the 1st rank student

```python
#  Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
print(marks.loc["Rank1":"Rank3"])
```

loc is label-based indexing.
It accesses data using the index names (Rank1, Rank2, Rank3).

```python
#  Use the iloc accessor to retrieve the mark of the 3rd rank student.
print(marks.iloc[2])
```

iloc is position-based indexing.
Position counting starts from 0.
So iloc[2] retrieves the mark of the 3rd rank student.

```python
#  Use the loc accessor to retrieve the marks of the last two ranks.
print(marks.loc["Rank4":"Rank5"])
```

```python
#  Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print(marks[marks>90])
```

```python
# 3. Manipulating Series:
#  Modify the mark of the student with the 1st rank to 100.
marks["Rank1"] = 100
print(marks)
```

replaces 1st rank mark with 100

```python
#  Add a new rank with a mark of 87.
marks["Rank6"] = 87
print(marks)
```

adds new element

```python
#  Remove the entry corresponding to the last rank from the Series.
marks = marks.drop("Rank5")
print(marks)
```

drops rank 5

```python
#  Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks/10
print(cgpa)
```

## ***Pandas DataFrame:***

A two-dimensional labeled data structure with
columns of potentially different types.
* Tabular structure (rows + columns)
* Flexible indexing
* Supports heterogeneous data types

```python
# 1. Creating Pandas DataFrame:
#  Create a Pandas DataFrame named transactions with the following data:
#  TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
#  ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
#  Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'
#  Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

Transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region": ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(Transactions)
```

```python
# 2. Data Exploration:
#  Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.
print(Transactions.head())
print(Transactions.tail())
print(Transactions.shape)
print(Transactions.columns)
print(Transactions.dtypes)
```

```python
#  Display only the 'ProductCategory' and 'Amount' columns.
print(Transactions[["ProductCategory", "Amount"]])
```

```python
#  Retrieve the last 3 columns of the DataFrame using iloc or loc.
print(Transactions.iloc[:, -3:])
```

iloc is position-based indexing.

```python
#  Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
print(Transactions[(Transactions["Region"] == "North") & (Transactions["Amount"] > 200)])
```

```python
#  Find the value counts for the 'ProductCategory' column.
print(Transactions["ProductCategory"].value_counts())
```

counts the productcategory column

```python
#  Find the unique values in the 'Region' column.
print(Transactions["Region"].unique())
```

```python
#  Group by 'Region' and find the mean amount for each region.
print(Transactions.groupby("Region")["Amount"].mean())
```

```python
# 3. Manipulating the DataFrame:
#  Modify the 'Amount' for the transaction with TransactionID 102 to 165.
Transactions.loc[Transactions["TransactionID"] == 102, "Amount"] = 165
print(Transactions)
```

```python
#  Add a new column 'Discount' by calculating 10% of the 'Amount'.
Transactions["Discount"] = Transactions["Amount"] * 0.1
print(Transactions)
```

```python
#  Remove the row with TransactionID 109.
Transactions = Transactions[Transactions["TransactionID"] != 109]
print(Transactions)
```

```python
#  Delete the 'Discount' column from the DataFrame.
del Transactions["Discount"]
print(Transactions)
```
